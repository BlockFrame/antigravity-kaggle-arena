```python
import json, shutil, subprocess, torch
smi = subprocess.run(['nvidia-smi', '--query-gpu=name,memory.total', '--format=csv,noheader'], capture_output=True, text=True) if shutil.which('nvidia-smi') else None
diagnostics = {'torch': torch.__version__, 'torch_cuda': torch.version.cuda, 'cuda_available': torch.cuda.is_available(), 'device_count': torch.cuda.device_count(), 'nvidia_smi': smi.stdout.strip() if smi else 'missing'}
print(json.dumps(diagnostics, indent=2))
assert diagnostics['cuda_available'], diagnostics
```
